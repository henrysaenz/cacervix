## Explicación de variables

- **COD_DPTO**: Departamento donde ocurrió la defunción  
- **COD_MUNIC**: Municipio donde ocurrió la defunción  
- **A_DEFUN**: Área donde ocurrió la defunción  
- **SIT_DEFUN**: Sitio de la defunción  
- **COD_INST**: Código de la institución prestadora de salud  
- **NOM_INST**: Nombre de la institución de salud  
- **ANO**: Año en que ocurrió la defunción  
- **MES**: Mes en que ocurrió la defunción  
- **SEXO**: Sexo del fallecido  
- **EST_CIVIL**: Estado conyugal del fallecido  
- **gru_ed1**: Agrupación de edades, según la edad del fallecido  
- **gru_ed2**: Agrupación alternativa de edades del fallecido  
- **NIVEL_EDU**: Nivel educativo alcanzado por el fallecido  
- **MUERTEPORO**: Ocupación como posible causa o factor asociado a la defunción  
- **SIMUERTEPO**: Si fue accidente de trabajo o enfermedad profesional  
- **OCUPACION**: Última ocupación habitual del fallecido  
- **IDPERTET**: Autoidentificación étnica del fallecido  
- **CODPRES**: País de residencia habitual si era diferente a Colombia  
- **CODPTORE**: Departamento de residencia habitual del fallecido  
- **CODMUNRE**: Municipio de residencia habitual del fallecido  
- **AREA_RES**: Área de residencia habitual del fallecido  
- **SEG_SOCIAL**: Régimen de seguridad social del fallecido  
- **IDADMISALU**: Entidad administradora en salud (o de la madre, si es muerte fetal o de menor de un año)  
- **IDCLASADMI**: Nombre de la administradora de salud  
- **pman_muer**: Probable manera de muerte  
- **CONS_EXP**: Certificación médica de la defunción  
- **CODOCUR**: Departamento de ocurrencia del hecho violento (si aplica)  
- **CODMUNOC**: Municipio de ocurrencia del hecho violento (si aplica)  
- **c_muerte**: Método de determinación de la causa de muerte  
- **ASIS_MED**: Si recibió asistencia médica antes de morir  
- **CAUSA_MULT**: Códigos de causas múltiples como antecedentes  
- **C_DIR1**: Causa directa de la defunción  
- **C_DIR12**: Segundo código informado para la causa directa  
- **C_ANT1**: Primera causa antecedente  
- **C_ANT12**: Segundo código informado para la causa antecedente 1  
- **C_ANT2**: Segunda causa antecedente  
- **C_ANT22**: Segundo código informado para la causa antecedente 2  
- **C_ANT3**: Tercera causa antecedente  
- **C_ANT32**: Segundo código informado para la causa antecedente 3  
- **C_PAT1**: Otros estados patológicos importantes  
- **C_PAT2**: Código correspondiente a C_PAT1  
- **C_BAS1**: Código de la causa básica de defunción  
- **C_MCM1**: Código de causa sin certificación médica  
- **IDPROFCER**: Profesión de quien certifica la defunción  
- **CAUSA_666_667**: Agrupación de causa según Lista 6/67 OPS  
- **CAU_HOMOL**: Agrupación de causa básica según Lista 105 Colombia  
- **QUINQ**: Quinquenio de edad del fallecido (variable agrupada en intervalos de 5 años)  
- **EDAD2**: Edad del fallecido en una unidad específica (días, meses o años, según contexto)

```python
def resumen_datos(data):
    resumen = []
    total_filas = len(data)
    
    for col in data.columns:
        unicos = data[col].nunique()
        faltantes = data[col].isna().sum()
        ceros = (data[col] == 0).sum()
        porcentaje_faltante = (faltantes / total_filas) * 100
        super_10 = 'Sí' if porcentaje_faltante > 10 else 'No'
        
        resumen.append({
            'Columna': col,
            'Valores únicos': unicos,
            'Faltantes': faltantes,
            'Ceros': ceros,
            'Porcentaje de faltantes': porcentaje_faltante,
            'Más del 10% de faltantes': super_10
        })
    return pd.DataFrame(resumen)
        
```

```python
# Distribución conocida (sin valores faltantes)
estado_civil_counts = data['ESTADO_CIVIL_IMPUTADO_MODA'].value_counts()
total_known = estado_civil_counts.sum()

# Calcular proporciones
proporciones = estado_civil_counts / total_known

# Cantidad de valores faltantes
n_missing = data['ESTADO_CIVIL_IMPUTADO_MODA'].isna().sum()

# Determinar cuántos valores imputar por categoría
asignaciones = (proporciones * n_missing).round().astype(int)
```

Imputación para la categoría `Sin información` de `área de residencia`

Se realizó una reclasificación de los registros con valor 'Sin informacion' en la variable AREA DE RESIDENCIA, utilizando como referencia la variable AREA_DEFUN. Para los casos en que AREA_DEFUN era 'URBANA', se asignó directamente la categoría 'Cabecera municipal'. Para los casos con AREA_DEFUN igual a 'RURAL', se redistribuyeron proporcionalmente entre 'Centro poblado (Inspección, corregimiento o caserío)' y 'Rural disperso', según la frecuencia observada de estas categorías en los datos. Con este procedimiento se eliminó la categoría ambigua, mejorando la calidad y precisión de la variable AREA DE RESIDENCIA.

```python
# Frecuencias de las opciones rurales conocidas
rural_counts = data['AREA DE RESIDENCIA'].value_counts()
centro_poblado = rural_counts['Centro poblado (Inspección, corregimiento o caserío)']
rural_disperso = rural_counts['Rural disperso']
total_rural = centro_poblado + rural_disperso
```

```python
p_centro = centro_poblado / total_rural
p_disperso = rural_disperso / total_rural

# Filtrar solo las filas con 'Sin informacion' en AREA DE RESIDENCIA
sin_info_mask = data['AREA DE RESIDENCIA'] == 'Sin informacion'
sin_info_indices = data[sin_info_mask].index
```

```python
np.random.seed(42) 

for idx in sin_info_indices:
    if data.at[idx, 'AREA_DEFUN'] == 'URBANA':
        data.at[idx, 'AREA DE RESIDENCIA'] = 'Cabecera municipal'
    else:  # RURAL
        # Asignar aleatoriamente según proporciones
        data.at[idx, 'AREA DE RESIDENCIA'] = np.random.choice(
            ['Centro poblado (Inspección, corregimiento o caserío)', 'Rural disperso'],
            p=[p_centro, p_disperso]
        )
```
