## Tasa de mortalidad por año

`````{admonition} **Tasa de mortalidad**
:class: tip
La **tasa de mortalidad** es una medida epidemiológica que indica la frecuencia de muertes por una causa específica en una población durante un periodo determinado.

$$
\text{Tasa de mortalidad} = \frac{\text{Número de defunciones por una causa específica en un periodo}}{\text{Población total expuesta al riesgo en el mismo periodo}} \times 100,000
$$

*La tasa suele expresarse por cada 100,000 habitantes.*
`````
